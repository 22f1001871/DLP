Assume that we want to continue pertaining a model containing 7 billion parameters with a context length of 2048 and the embedding dimension (model dimension) of 2048. The dataset contains 1 million samples (the size of each sample is exactly equal to the context length of the model). Suppose we set “per device batch size” to 16 and the gradient accumulation steps to 4.

1.How many optimizer update steps are required to complete one epoch on a single GPU?

```python
total_param = 1000000
#no of samples the CPU handles  before the gradient accumulation pass
per_device_batch_size = 16
gradient_accumulation_step = 4

effective_batch_size = per_device_batch_size * gradient_accumulation_step

effective_param = total_param / effective_batch_size

print("No of optimizer update steps required to complete one epoch on a single GPU",effective_param)
```

2. The statement that the weight update happens only at the end of one epoch is

False. Because, here one epoch is 16 batches. But the gradient update happens for every batch

3. How much memory (GB) is required to load the model with fp32? Use 1GB = 1e9.

```python
parameters = 7*10**9
#bytes per parameter
fp32_precision = 4

total_bytes = parameters*fp32_precision

#GB
print("The memory required is",total_bytes/10**9)
```

4. Suppose the context length is increased from 2048 to 8000. Then, how much memory (GB) is required to load the model with fp32? Use 1GB = 1e9.

```python
#Now the parameter increases by 2048*8000

parameters = (2048*8000) - (2048*2048)
updated_param = 7*10**9 + parameters


fp32_precision = 4

total_bytes = updated_param*fp32_precision

#GB
print("The memory required is",total_bytes/10**9)
```
