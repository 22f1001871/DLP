## 1. Download the BookCorpus dataset. Take every 7-th sample (the indices are multiple of 7:[0,7,14,21,...]) from the entire dataset. This will result in a dataset with 10 million samples (exactly, 10,572,033). Use these samples to build a tokenizer with the BPE tokenization algorithm by varying the vocabulary size.

    Normalizer: LowerCase
    PreTokenizer: WhiteSpace
    Model: BPE
    Special tokens: [GO],[UNK],[PAD],[EOS]
    PostProcessing: None


Tokenize the input text: “SEBI study finds 93% of individual F&O traders made losses between FY22 and FY24.” using the following configurations

```python
from datasets import load_dataset

data = load_dataset("rojagtap/bookcorpus",split="all")
```

```python
print("Total datasets after loading the dataset",len(data))
```

```python
df = data.filter(lambda example,  idx  : idx%7 == 0, with_indices=True)
```

```python
len(df)
```

```python
#Importing required libraries
from tokenizers import Tokenizer
from tokenizers.normalizers import Lowercase
from tokenizers.pre_tokenizers import Whitespace
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer

model = BPE(unk_token="[UNK]")
tokenizer_5k = Tokenizer(model)

tokenizer_5k.normalizer = Lowercase()
tokenizer_5k.pre_tokenizer = Whitespace()

trainer = BpeTrainer(vocab_size=5000, special_tokens=["[GO]","[UNK]","[PAD]","[EOS]"], continuing_subword_prefix="##")

tokenizer_5k.train_from_iterator(get_samples(), trainer=trainer)

```

```python
#to get batch of samples
def get_samples(batch_size=1000):
  n=len(df)
  for i in range (0,n, batch_size):
   yield df[i:i+batch_size]['text']
```

```python
tokenizer_5k.get_vocab_size()
```

```python
sample ="SEBI study finds 93% of individual F&O traders made losses between FY22 and FY24."
print(f'sample: {sample}')
encoding = tokenizer_5k.encode(sample)
print(f'tokens: {encoding.tokens}')
print(f'ids: {encoding.ids}')
print("Total no of tokens",len(encoding.tokens))
```

## 7. By now, we have four tokenizers.

1. Custom tokenizer (vocab size 32K, trained on 10 million samples)
2. bert-base-uncased
3. gpt2
4. hopper

Use these four tokenizers to count the number of tokens for the entire “imdb” dataset (drop the “unsupervised” part of the dataset). Enter the tokenizers in order such that the size of the dataset (measured in tokens) as returned by the tokenizers is in ascending order. For example, if the first tokenizer yields the smallest number of tokens and the fourth tokenizer yields the largest, you would enter 1234 (without any spaces).”

```python
from tokenizers import Tokenizer
tokenizer_32k = Tokenizer.from_file("/content/tokenizer_32k.json")

hopper = Tokenizer.from_file("/content/hopper.json")
```

```python
from datasets import load_dataset
data = load_dataset("stanfordnlp/imdb")
```

```python
from datasets import concatenate_datasets
df = concatenate_datasets([data['train'],data['test']])
```

```python
df.remove_columns(["label"])
```

```python
total_32 = 0
total_hopper = 0
total_bert = 0
total_gpt = 0
for i in enumerate(df['text']):
  try:
    token_32k=tokenizer_32k.encode(i[1])
    token_hopper=hopper.encode(i[1])
    token_bert=bert_tokenizer.encode(i[1])
    token_gpt=gpt.encode(i[1])
    total_32 += len(token_32k.tokens)
    total_hopper += len(token_hopper.tokens)
    total_bert += len(token_bert)
    total_gpt += len(token_gpt)
  except:
    print(i[1])
```

```python
a=[['1',total_32],['2',total_bert],['3',total_gpt],['4',total_hopper]]
```

```python
a
```

```python
i=0
while i<len(a):
  j=0
  while j<len(a):
    if a[i][1]<a[j][1]:
      b=a[i]
      c=a[j]
      a[i]=c
      a[j]=b
    j+=1
  i+=1
```
